## Introduction

Real-world datasets are rarely perfect. They often contain missing values, duplicate records, inconsistent formats, incorrect data types, invalid entries, and other quality issues that can affect analysis and machine learning models.

Before performing exploratory data analysis (EDA), it is essential to thoroughly understand the dataset and improve its quality through systematic preprocessing.

The objective of this notebook is to build a robust and reusable data preprocessing pipeline capable of handling common data quality problems automatically. Rather than cleaning only the current dataset, the workflow is designed to adapt to future datasets with similar structures but different data patterns.

This notebook is divided into two major phases:

### Phase 1 — Data Understanding

The first phase focuses on understanding the dataset by examining:

- Dataset dimensions
- Feature information
- Data types
- Missing values
- Duplicate records
- Memory usage
- Statistical summaries
- Potential data quality issues

Understanding the data before modification helps identify problems and ensures informed preprocessing decisions.

### Phase 2 — Data Cleaning

The second phase performs comprehensive preprocessing to improve data quality by handling:

- Missing values
- Duplicate records
- Incorrect data types
- Invalid categorical values
- Inconsistent text formatting
- Date and time conversion
- Boolean standardization
- Numeric conversion
- Outlier detection
- Data validation
- Feature consistency checks

The resulting dataset becomes cleaner, more consistent, and better suited for exploratory data analysis, visualization, and predictive modeling.

## Load Dataset

The dataset is loaded into a pandas DataFrame for further inspection and preprocessing. After loading, the first few records are displayed to verify that the data has been imported correctly.

```python
df = pd.read_csv("/kaggle/input/datasets/satyakidas07/cityflow-bus-service-metro-cities/cityflo_bus_service_metro_cities.csv")
df.head()
```

## Basic Dataset Information

Before cleaning the data, it is important to understand its overall structure. This includes the number of observations, number of features, memory usage, and data types stored within the dataset.

These details help identify potential preprocessing requirements such as datatype conversion, missing values, or memory optimization.

```python
print(f"Number of Rows    : {df.shape[0]}")
print(f"Number of Columns : {df.shape[1]}")

print("\nDataset Shape:")
print(df.shape)

print("\nMemory Usage (MB):")
print(round(df.memory_usage(deep=True).sum() / (1024 ** 2), 2), "MB")
```

## Dataset Overview

A detailed understanding of each feature is essential before applying any preprocessing techniques. This section provides comprehensive information about every column, including its data type, missing values, number of unique values, memory consumption, and sample values.

The overview helps identify potential issues such as:

- Incorrect data types
- Missing values
- High-cardinality features
- Constant columns
- Identifier columns
- Features requiring encoding
- Features requiring standardization

```python
print("DATASET INFORMATION")
df.info()
```

## Column-wise Metadata

To better understand the characteristics of each feature, a metadata summary is generated. This table provides important information about every column, including:

- Data type
- Missing values
- Missing percentage
- Number of unique values
- Memory usage
- Sample values

This detailed overview assists in planning appropriate preprocessing steps for each feature.

```python
column_summary = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.values,
    "Non-Null Count": df.count().values,
    "Missing Values": df.isnull().sum().values,
    "Missing (%)": (
        (df.isnull().sum() / len(df)) * 100
    ).round(2).values,
    "Unique Values": df.nunique().values,
    "Memory (KB)": (
        df.memory_usage(deep=True)[1:] / 1024
    ).round(2).values
})

column_summary
```

## Sample Values

Inspecting representative values from each feature provides an initial understanding of the data stored within the column. It also helps identify formatting inconsistencies, unexpected symbols, abbreviations, and possible data entry errors.

```python
sample_values = pd.DataFrame({
    "Column": df.columns,
    "Sample Values": [
        ", ".join(map(str, df[col].dropna().unique()[:5]))
        for col in df.columns
    ]
})

sample_values
```

## Statistical Summary of Numerical Features

Numerical variables are summarized using descriptive statistics such as:

- Count
- Mean
- Standard deviation
- Minimum

- Maximum

These statistics help identify unusual distributions, potential outliers, and impossible values that may require preprocessing.

## Statistical Summary of Categorical Features

Categorical variables are summarized to identify:

- Total observations
- Number of unique categories
- Most frequent category
- Frequency of the most common category

Understanding categorical features helps identify inconsistent spellings, rare categories, and potential standardization requirements.

```python
categorical_summary = (
    df.select_dtypes(include="object")
      .describe()
      .T
)

categorical_summary
```

## Automatic Feature Categorization

To make the preprocessing pipeline adaptable to future datasets, features are automatically categorized based on their data types.

This dynamic approach ensures that preprocessing steps remain applicable even if additional columns are introduced or existing columns are modified.

```python
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

categorical_cols = df.select_dtypes(include="object").columns.tolist()

datetime_cols = df.select_dtypes(include=["datetime64[ns]"]).columns.tolist()

print(f"Numerical Features   : {len(numerical_cols)}")
print(numerical_cols)

print("\nCategorical Features :", len(categorical_cols))
print(categorical_cols)

print("\nDatetime Features    :", len(datetime_cols))
print(datetime_cols)
```

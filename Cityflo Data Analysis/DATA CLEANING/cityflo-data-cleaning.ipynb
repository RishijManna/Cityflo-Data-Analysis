## Quasi-Constant Feature Detection

A quasi-constant feature contains one dominant value appearing in the vast majority of observations (typically more than 95%).

Although not completely constant, these features often contribute very little information and may be considered for removal depending on the application.

```python
threshold = 0.95

quasi_constant = []

for col in df.columns:

    dominant = (
        df[col]
        .value_counts(normalize=True,
                      dropna=False)
        .max()
    )

    if dominant >= threshold:
        quasi_constant.append({
            "Column": col,
            "Dominant Ratio": round(dominant,4)
        })

pd.DataFrame(quasi_constant)
```

## Standardize Column Names

Datasets collected from different sources often contain inconsistent column names due to varying capitalization, spaces, special characters, or formatting styles.

To improve readability and ensure compatibility with Python programming conventions, column names are standardized by:

- Removing leading and trailing spaces
- Converting all names to lowercase
- Replacing spaces with underscores
- Replacing hyphens with underscores
- Removing unnecessary special characters
- Eliminating duplicate underscores

This standardization improves code readability and reduces the likelihood of errors during analysis.

```python
import re

original_columns = df.columns.tolist()

clean_columns = []

for col in df.columns:

    col = str(col)

    # Remove leading/trailing spaces
    col = col.strip()

    # Convert to lowercase
    col = col.lower()

    # Replace spaces and hyphens with underscores
    col = re.sub(r"[\s\-]+", "_", col)

    # Remove special characters except underscore
    col = re.sub(r"[^a-z0-9_]", "", col)

    # Replace multiple underscores
    col = re.sub(r"_+", "_", col)

    # Remove leading/trailing underscores
    col = col.strip("_")

    clean_columns.append(col)

# Handle duplicate column names
seen = {}
final_columns = []

for col in clean_columns:
    if col not in seen:
        seen[col] = 0
        final_columns.append(col)
    else:
        seen[col] += 1
        final_columns.append(f"{col}_{seen[col]}")

df.columns = final_columns

print("Column names standardized successfully.")
```

## Automatic Numeric Conversion

Numeric information is sometimes imported as text because of inconsistent formatting, delimiters, or missing values.

Instead of manually specifying column names, the preprocessing pipeline automatically identifies object-type columns that predominantly contain numeric values and converts them into the appropriate numeric data type.

This dynamic approach improves reusability across future datasets without requiring code modifications.

```python
converted_numeric = []

for col in df.select_dtypes(include="object").columns:

    converted = pd.to_numeric(df[col], errors="coerce")

    non_null = df[col].notna().sum()

    if non_null == 0:
        continue

    success_rate = converted.notna().sum() / non_null

    if success_rate >= 0.95:

        df[col] = converted

        converted_numeric.append(col)


print("NUMERIC CONVERSION")


if converted_numeric:

    print("Converted Columns:\n")

    for col in converted_numeric:
        print(f"• {col}")

else:
    print("No additional numeric columns detected.")
```

## Automatic Date Detection

Dates are frequently stored as text, preventing chronological analysis and time-based feature engineering.

The preprocessing pipeline automatically detects columns that predominantly contain valid date values and converts them into datetime format while preserving invalid entries as missing values.

```python
converted_dates = []

for col in df.select_dtypes(include="object").columns:

    converted = pd.to_datetime(
        df[col],
        errors="coerce",
        infer_datetime_format=True
    )

    non_null = df[col].notna().sum()

    if non_null == 0:
        continue

    success_rate = converted.notna().sum() / non_null

    if success_rate >= 0.95:

        df[col] = converted

        converted_dates.append(col)

print("DATE CONVERSION")

if converted_dates:

    print("Converted Date Columns:\n")

    for col in converted_dates:
        print(f"• {col}")

else:
    print("No date columns detected.")
```

## Boolean Value Standardization

Binary variables are often represented using different conventions such as **Yes/No**, **True/False**, **Y/N**, or **1/0**. Treating these representations consistently improves data quality and simplifies downstream analysis.

The preprocessing pipeline automatically detects common binary representations and converts them into standard boolean values.

```python
boolean_mapping = {
    "yes": True,
    "no": False,
    "true": True,
    "false": False,
    "y": True,
    "n": False,
    "1": True,
    "0": False
}

boolean_columns = []

for col in df.select_dtypes(include="object").columns:

    values = (
        df[col]
        .dropna()
        .astype(str)
        .str.lower()
        .unique()
    )

    if len(values) <= 2:

        if set(values).issubset(boolean_mapping.keys()):

            df[col] = (
                df[col]
                .astype(str)
                .str.lower()
                .map(boolean_mapping)
            )

            boolean_columns.append(col)

print("BOOLEAN STANDARDIZATION")
if boolean_columns:

    print("Boolean Columns Converted:\n")

    for col in boolean_columns:
        print(f"• {col}")

else:
    print("No boolean columns detected.")
```

## Missing Value Imputation

Missing values can negatively impact statistical analysis and machine learning algorithms. Since different types of variables have different characteristics, a single imputation strategy is not suitable for every feature.

The preprocessing pipeline applies type-specific imputation strategies:

- **Numerical Features:** Missing values are replaced using the median to reduce the influence of extreme values.
- **Categorical Features:** Missing values are replaced using the most frequent category (mode).
- **Datetime Features:** Missing timestamps are filled using forward filling followed by backward filling to preserve chronological consistency whenever possible.

This adaptive approach ensures that the dataset remains complete while minimizing distortion of the original data distribution.

```python

imputation_summary = []

for col in df.columns:

    missing_before = df[col].isnull().sum()

    if missing_before == 0:
        continue

    # Numerical Columns
    if pd.api.types.is_numeric_dtype(df[col]):

        median = df[col].median()

        df[col].fillna(median, inplace=True)

        strategy = "Median"

    # Datetime Columns
    elif pd.api.types.is_datetime64_any_dtype(df[col]):

        df[col] = df[col].ffill().bfill()

        strategy = "Forward Fill + Backward Fill"

    # Categorical Columns
    else:

        mode = df[col].mode(dropna=True)

        if not mode.empty:
            df[col].fillna(mode.iloc[0], inplace=True)
        else:
            df[col].fillna("Unknown", inplace=True)

        strategy = "Mode"

    imputation_summary.append({
        "Column": col,
        "Strategy": strategy,
        "Values Imputed": missing_before
    })

display(pd.DataFrame(imputation_summary))
```

```python
range_summary = pd.DataFrame({
    "Minimum": df.select_dtypes(include=np.number).min(),
    "Maximum": df.select_dtypes(include=np.number).max(),
    "Mean": df.select_dtypes(include=np.number).mean().round(2),
    "Median": df.select_dtypes(include=np.number).median().round(2)
})

display(range_summary)
```

## Impossible Value Detection

Certain numerical features have logical constraints. For example, travel distance, fare amount, passenger count, and delay duration should not contain negative values.

The preprocessing pipeline automatically checks for impossible values in commonly encountered transportation-related attributes and reports them for further review.

```python
rules = {
    "fare": lambda x: x < 0,
    "distance": lambda x: x < 0,
    "passenger": lambda x: x <= 0,
    "delay": lambda x: x < 0,
    "rating": lambda x: (x < 0) | (x > 5),
    "duration": lambda x: x <= 0
}

report = []

for col in df.select_dtypes(include=np.number).columns:

    for keyword, condition in rules.items():

        if keyword in col.lower():

            invalid = condition(df[col]).sum()

            if invalid > 0:

                report.append({
                    "Column": col,
                    "Invalid Records": int(invalid)
                })

if report:
    display(pd.DataFrame(report))
else:
    print("✅ No impossible values detected based on predefined rules.")
```

## Outlier Detection Using the Interquartile Range (IQR)

Outliers are observations that deviate significantly from the majority of the data. They may arise due to measurement errors, data entry mistakes, or naturally occurring extreme events.

The Interquartile Range (IQR) method identifies potential outliers without making assumptions about the underlying data distribution.

This step detects potential outliers but does not remove them, allowing informed decisions during subsequent analysis.

```python
# =====================================================
# IQR Outlier Detection
# =====================================================

outlier_summary = []

numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        "Column": col,
        "Outliers": outliers,
        "Outlier (%)": round(outliers / len(df) * 100, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary)

display(outlier_summary.sort_values("Outliers", ascending=False))
```

## Z-Score Based Outlier Verification

The Z-score method measures how many standard deviations each observation lies from the mean.

Values with an absolute Z-score greater than three are commonly regarded as potential outliers.

This step complements the IQR method by providing an additional perspective on extreme observations.

```python
zscore_summary = []

for col in numeric_columns:

    if df[col].std() == 0:
        continue

    z_scores = np.abs(stats.zscore(df[col]))

    count = (z_scores > 3).sum()

    zscore_summary.append({
        "Column": col,
        "Extreme Values": int(count)
    })

display(pd.DataFrame(zscore_summary))
```

## Rare Category Detection

Categories appearing only a few times may represent genuine but infrequent observations, typographical errors, or inconsistent labeling.

Identifying these categories helps determine whether they should be merged, corrected, or retained.

```python
# =====================================================
# Rare Category Detection
# =====================================================

rare_limit = 5

rare_categories = {}

for col in df.select_dtypes(include="object").columns:

    rare = df[col].value_counts()

    rare = rare[rare <= rare_limit]

    if not rare.empty:
        rare_categories[col] = rare

for col, values in rare_categories.items():

    print("=" * 70)
    print(col.upper())
    print("=" * 70)

    display(values.to_frame("Frequency"))
```
